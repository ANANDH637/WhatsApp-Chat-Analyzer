```python
# WhatsApp Chat Analyzer

import re
from collections import Counter

chat = """
01/10/2026, 09:15 - Akshaya: Good morning everyone!
01/10/2026, 09:20 - Rahul: Good morning!
01/10/2026, 10:05 - Akshaya: Did you complete the project?
01/10/2026, 10:10 - Priya: I completed my part.
01/10/2026, 10:15 - Rahul: I am working on it now.
01/10/2026, 11:00 - Akshaya: Great! Please share the files.
01/10/2026, 11:20 - Priya: Sure, I will send them.
01/10/2026, 12:05 - Rahul: The project looks good!
"""

# Extract date, time, sender and message
pattern = r'(\d{2}/\d{2}/\d{4}),\s(\d{2}:\d{2})\s-\s([^:]+):\s(.+)'

messages = re.findall(pattern, chat)

senders = []
message_texts = []

for date, time, sender, message in messages:
    senders.append(sender.strip())
    message_texts.append(message.strip())

# Message statistics
message_count = Counter(senders)

# Word analysis
all_words = []

for message in message_texts:
    words = re.findall(r'\b[a-zA-Z]+\b', message.lower())
    all_words.extend(words)

stop_words = {
    "the", "is", "am", "are", "a", "an", "i",
    "to", "it", "my", "on", "and", "will",
    "you", "your", "did", "them", "now"
}

meaningful_words = [
    word for word in all_words
    if word not in stop_words
]

word_frequency = Counter(meaningful_words)

# Questions and exclamations
question_count = sum("?" in msg for msg in message_texts)
exclamation_count = sum("!" in msg for msg in message_texts)

# Sentiment
positive_words = {
    "good", "great", "completed", "sure", "nice",
    "excellent", "happy", "looks"
}

negative_words = {
    "bad", "sad", "angry", "problem", "worst"
}

positive_count = sum(
    1 for word in all_words if word in positive_words
)

negative_count = sum(
    1 for word in all_words if word in negative_words
)

if positive_count > negative_count:
    sentiment = "Positive"
elif negative_count > positive_count:
    sentiment = "Negative"
else:
    sentiment = "Neutral"


print("=" * 60)
print("               WHATSAPP CHAT ANALYZER")
print("=" * 60)

print("\nTotal Messages :", len(messages))
print("Participants   :", len(message_count))
print("Questions      :", question_count)
print("Exclamations   :", exclamation_count)

print("\nMessages Per Person:")

for sender, count in message_count.items():
    print(f"  {sender}: {count}")

most_active = message_count.most_common(1)[0]

print("\nMost Active Participant:")
print(f"  {most_active[0]} ({most_active[1]} messages)")

print("\nTop 5 Words:")

for word, count in word_frequency.most_common(5):
    print(f"  {word}: {count}")

print("\nChat Sentiment:")
print("  Positive Words :", positive_count)
print("  Negative Words :", negative_count)
print("  Overall Tone   :", sentiment)

print("\n" + "=" * 60)
print("WhatsApp Chat Analysis Completed!")
print("=" * 60)
```
